```python
a = int(input())
b = int(input())
c = int(input())
if a+b>c or b+c>a or a+c>b:
  if a==b and b==c:
    print("equilateral")
  elif a==b or b==c or a==c:
    print("isosceles")
  else:
    print("scalene")
else:
  print("not a valid triangle")
```
